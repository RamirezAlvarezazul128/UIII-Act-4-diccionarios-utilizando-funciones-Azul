#### DICCIONARIO DE BAILARIN

```python
# --- PROGRAMA DE GESTIÓN DE BAILARINES (SISTEMA INTERACTIVO) ---

# GigaNota: Inicio con una lista de diccionarios vacía [13, 14]
lista_bailarines = []

def Listar_bailarines():
    """Muestra la información en forma de tabla bien distribuida [MegaNota]"""
    if not lista_bailarines:
        print("\n[!] La lista de bailarines está vacía.")
    else:
        print("\n" + "="*117)
        # Encabezado con tamaños fijos exactos: 20, 15, 15, 20, 15, 15
        print(f"{'Compañía':<20} | {'Estilo':<15} | {'Flexibilidad':<15} | {'Coreografías':<20} | {'Calzado':<15} | {'Premios':<15}")
        print("-" * 117)

        # Uso de ciclo for para recorrer la lista de diccionarios [9, 15]
        for bailarin in lista_bailarines:
            # Imprime cada dato usando exactamente el mismo espacio que el encabezado
            print(f"{bailarin['Compañía']:<20} | {bailarin['Estilo']:<15} | {bailarin['Nivel']:<15} | "
                  f"{bailarin['Coreografías']:<20} | {bailarin['Calzado']:<15} | {bailarin['Premios']:<15}")
        print("="*117 + "\n")

def agregar_bailarin():
    """Agrega un nuevo diccionario de bailarín a la lista"""
    print("\n--- Registro de Nuevo Bailarín ---")
    nuevo = {}
    nuevo['Compañía'] = input("Nombre de la compañía: ")
    nuevo['Estilo'] = input("Estilo (Ballet/Contemporáneo): ")
    nuevo['Nivel'] = input("Nivel de flexibilidad: ")
    nuevo['Coreografías'] = input("Coreografías destacadas: ")
    nuevo['Calzado'] = input("Calzado utilizado: ")
    nuevo['Premios'] = input("Premios obtenidos: ")

    lista_bailarines.append(nuevo)
    print("\n[+] Bailarín agregado exitosamente.")
    Listar_bailarines() # Siempre muestra la tabla al finalizar

def buscar_bailarin():
    """Busca un bailarín por el nombre de la compañía"""
    print("\n--- Buscar Bailarín ---")
    criterio = input("Ingrese el nombre de la Compañía a buscar: ")
    encontrado = False

    for bailarin in lista_bailarines:
        if bailarin['Compañía'].lower() == criterio.lower():
            print("\n[ Resultado de la búsqueda ]")
            for clave, valor in bailarin.items():
                print(f"{clave}: {valor}")
            encontrado = True
            break

    if not encontrado:
        print("\n[!] No se encontró ninguna compañía con ese nombre.")
    Listar_bailarines()

def borrar_bailarin():
    """Elimina un bailarín de la lista"""
    print("\n--- Borrar Registro ---")
    criterio = input("Ingrese el nombre de la Compañía a eliminar: ")

    for i in range(len(lista_bailarines)):
        if lista_bailarines[i]['Compañía'].lower() == criterio.lower():
            eliminado = lista_bailarines.pop(i)
            print(f"\n[-] Registro de '{eliminado['Compañía']}' borrado.")
            break
    else:
        print("\n[!] No se encontró el registro para eliminar.")
    Listar_bailarines()

def actualizar_bailarin():
    """Modifica los datos de un bailarín existente"""
    print("\n--- Actualizar Datos ---")
    criterio = input("Ingrese la Compañía del bailarín a actualizar: ")

    for bailarin in lista_bailarines:
        if bailarin['Compañía'].lower() == criterio.lower():
            print("Ingrese los nuevos datos (o presione Enter para mantener el actual):")
            nuevo_estilo = input(f"Estilo actual ({bailarin['Estilo']}): ")
            if nuevo_estilo: bailarin['Estilo'] = nuevo_estilo

            nuevo_premio = input(f"Premios actuales ({bailarin['Premios']}): ")
            if nuevo_premio: bailarin['Premios'] = nuevo_premio

            print("\n[*] Datos actualizados correctamente.")
            break
    else:
        print("\n[!] Compañía no encontrada.")
    Listar_bailarines()

# --- BUCLE PRINCIPAL (MENÚ INTERACTIVO) ---
while True:
    print("*"*40)
    print("   SISTEMA DE GESTIÓN: ARTES ESCÉNICAS")
    print("*"*40)
    print("1. Agregar bailarín")
    print("2. Listar diccionarios de bailarines")
    print("3. Buscar bailarín")
    print("4. Borrar bailarín")
    print("5. Actualizar bailarín")
    print("6. Salir")
    print("*"*40)

    opcion = input("Seleccione una opción (1-6): ")

    # Estructura if-elif para llamadas a funciones [3, 16]
    if opcion == '1':
        agregar_bailarin()
    elif opcion == '2':
        Listar_bailarines()
    elif opcion == '3':
        buscar_bailarin()
    elif opcion == '4':
        borrar_bailarin()
    elif opcion == '5':
        actualizar_bailarin()
    elif opcion == '6':
        print("\nCerrando el telón... ¡Hasta la próxima función!")
        break
    else:
        print("\n[!] Opción no válida. Intente de nuevo.")
```
